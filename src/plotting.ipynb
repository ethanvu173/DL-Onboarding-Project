```python
# UTAT Space Systems Deep Learning On-Boarding Project, Task 1
# Ethan Vu

import matplotlib.pyplot as plt
import csv
from pathlib import Path

n = 67  # HAHAHA
c = 0
wavelengths = []
refl = []
spectra = []
data_path = Path('../datasets/simpler_data_rwc.csv')

with open(data_path,'r') as finch_data:
    lines = csv.reader(finch_data, delimiter=',')
    for i, row in enumerate(lines, start=1):
        if i % n == 0:
            wavelengths.append([])
            refl.append([])
            for j in range(210):
                wavelengths[c].append(400 + 10 * j)
                refl[c].append(float(row[7+j]))
            spectra.append(row[0])
            c += 1

for i in range(25):
    plt.plot(wavelengths[i], refl[i], label=spectra[i], color=(i / 25, i / 25, 0.5))  # lowkey splatoon 3 colours

# plt.title('Wavelengths of Spectra')
plt.xlabel('Wavelength [nm]')
plt.ylabel('Reflectance')
plt.yticks([0, 0.2, 0.4, 0.6, 0.8, 1])

plt.show()
```
